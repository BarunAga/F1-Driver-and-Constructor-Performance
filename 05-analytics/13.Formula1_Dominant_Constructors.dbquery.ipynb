```sql
WITH dominant_constructors AS ( 

SELECT  constructor_name,
        SUM(total_points) AS total_points,
        SUM(number_of_wins) AS total_wins,
        SUM(number_of_podiums) AS total_podiums,
        SUM(race_start) AS race_number,
        SUM(CASE WHEN standing_position = 1 THEN 1 ELSE 0 END) AS total_championships
FROM formula1.gold.view_constructor_standings

GROUP BY constructor_name
HAVING total_championships > 0

)
SELECT  constructor_name,
        total_points,
        total_wins,
        total_podiums,
        race_number,
        total_championships,
        (100*total_championships + 7*total_wins + 3*total_podiums) AS greatness_score
FROM dominant_constructors
ORDER BY greatness_score DESC
```
